```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

Exercice 1 — Comprendre les pixels

```python
#charger l'image
image = Image.open("spider.jpg")

#convertir l'image en tableau numpy
image = np.array(image)

#afficher la forme du tableau numpy
print(image.shape)
```

```python
#afficher l'image
plt.imshow(image)
plt.show()
```

```python
#afficher la veleur RGB d'un pixel
pixel = image[100, 100, :3]
print(f"la valeur RGB du pixel (100, 100) est : {pixel}")
```

```python
#transformer ce pixel en rouge
image[0:100, 0:100, :3] = [255, 0, 0]
plt.imshow(image)
plt.show()
```

Exercice 2 — Convertir une image en niveaux de gris

```python
#charger l'image
image = Image.open("eye.jpg")

#convertir l'imageen tableau numpy
image = np.array(image)

#afficher la form du tableau numpy
print(image.shape)
```

```python
#afficher l'image
plt.imshow(image)
plt.show()
```

```python
#slicer une partie de l'image
R = image[:, 350:, 0]
G = image[:, 350:, 1]
B = image[:, 350:, 2]

#appliquer la formule de conversion en niveaux de gris
gris = 0.299 * R + 0.587 * G + 0.114 * B
print(gris)

#convertir le tableau en entier non signé de 8 bits
gris = gris.astype(np.uint8)

#print(gris.shape)
```

```python
#remplacer la partie de l'image par l'image en niveaux de gris
image[:, 350:, 0] = gris
image[:, 350:, 1] = gris
image[:, 350:, 2] = gris

#afficher l'image
plt.imshow(image)
plt.show()
```

Exercice 3 — Modifier la luminosité

```python
#charger l'image
image = Image.open("joker.jpg")

#convertir l'image en tableau numpy
image = np.array(image)

#afficher la forme du tableau
print(image.shape)
```

```python
#afficher l'image
plt.imshow(image)
plt.show()
```

```python
def modifier(img):
    #img = tableau numpy
    k = 100 #constante

    img_increase = img.astype(np.int16) + k
    img_increase = np.clip(img_increase, 0, 255).astype(np.uint8)

    img_decrease = img.astype(np.int16) - k
    img_decrease = np.clip(img_decrease, 0, 255).astype(np.uint8)

    increase, decrease = img_increase, img_decrease

    plt.figure()
    plt.imshow(increase)
    plt.title("Augmentation de la luminosité")
    plt.show() 
    plt.imshow(decrease)
    plt.title("Diminution de la luminosité")
    plt.show()


modifier(image)
```
